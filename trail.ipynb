```python
from pathlib import Path
import time

from mira_engine import MiraEngine, run_text_consultation, run_voice_consultation

CSV = Path("med.csv")
engine = MiraEngine(CSV)
print(f"Loaded {len(engine.df)} conditions from {CSV.resolve()}")

# Quick benchmark: symptom lookup
t0 = time.perf_counter()
for _ in range(500):
    engine.match_symptom("cough")
elapsed_ms = (time.perf_counter() - t0) * 1000 / 500
print(f"Avg symptom lookup: {elapsed_ms:.3f} ms")
```

```python
# Fastest: type your answers (recommended)
run_text_consultation(engine, use_voice=False)
```

```python
# Programmatic example (e.g. for widgets / API testing)
session = engine.new_session()
for msg in ["fever", "7", "moderate", "none"]:
    r = engine.process(session, msg)
    print(f"User: {msg}")
    print(f"MIRA: {r.message}")
    if r.done:
        break
```
